```python
daily_df["month"] = daily_df["date"].dt.to_period("M")
```

```python
monthly_data = daily_df.groupby("month").agg({
    "visitors": "sum",
    "product_viewers": "sum",
    "cart_users": "sum",
    "checkout_users": "sum",
    "purchasers": "sum"
}).reset_index()

monthly_data
```

```python
monthly_data["visitor_to_product_pct"] = (
    monthly_data["product_viewers"] / monthly_data["visitors"] * 100
)

monthly_data["product_to_cart_pct"] = (
    monthly_data["cart_users"] / monthly_data["product_viewers"] * 100
)

monthly_data["cart_to_checkout_pct"] = (
    monthly_data["checkout_users"] / monthly_data["cart_users"] * 100
)

monthly_data["checkout_to_purchase_pct"] = (
    monthly_data["purchasers"] / monthly_data["checkout_users"] * 100
)

monthly_data["overall_conversion_pct"] = (
    monthly_data["purchasers"] / monthly_data["visitors"] * 100
)

monthly_data
```

```python
best_month = monthly_data.loc[
    monthly_data["overall_conversion_pct"].idxmax()
]

worst_month = monthly_data.loc[
    monthly_data["overall_conversion_pct"].idxmin()
]

print("BEST MONTH")
print(best_month)

print("\nWORST MONTH")
print(worst_month)
```

```python
funnel_totals = {
    "Visitors": daily_df["visitors"].sum(),
    "Product Viewers": daily_df["product_viewers"].sum(),
    "Cart Users": daily_df["cart_users"].sum(),
    "Checkout Users": daily_df["checkout_users"].sum(),
    "Purchasers": daily_df["purchasers"].sum()
}

funnel_totals
```

```python
funnel_conversion = {
    "Visitor → Product": (
        funnel_totals["Product Viewers"]
        / funnel_totals["Visitors"] * 100
    ),

    "Product → Cart": (
        funnel_totals["Cart Users"]
        / funnel_totals["Product Viewers"] * 100
    ),

    "Cart → Checkout": (
        funnel_totals["Checkout Users"]
        / funnel_totals["Cart Users"] * 100
    ),

    "Checkout → Purchase": (
        funnel_totals["Purchasers"]
        / funnel_totals["Checkout Users"] * 100
    )
}

funnel_conversion
```
